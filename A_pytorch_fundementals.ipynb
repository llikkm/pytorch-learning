### Random Tensors

why random tensors?

Random tensors are important because many networks learn is that they start with tensors full of random numbers and then adjust those random numbers to better represent the data.

`Start with random numbers->look at data->update random numbers->look at data->update random numbers`

```python
#Create a random tensor of size(3,4)
random_tensor = torch.rand(2,3,4)
random_tensor
```

```python
random_tensor.ndim
```

```python
#create a tensor with simliar shape to an image tensor
random_image_size_tensor = torch.rand(size=(3,224,224))
# color channel(RGB),height, width
random_image_size_tensor#.shape, random_image_size_tensor.ndim

```

```python
random_tensor = torch.rand(size=(3,4,5))
random_tensor
```

### tensor datatypes

**note:** Tensor datatypes is one of 3 big errors you'll run into with pytorch & deep learning:
1. tensors not right datatype
2. tensors not right shape
3. tensors not on the right device


```python
#float 32
float_32_tensor = torch.tensor([3.0,6.0,9.0],
                               dtype=None,#datatype of the tensor(eg.float16,float32)
                               device="cuda",#
                               requires_grad=False)
float_32_tensor
```
