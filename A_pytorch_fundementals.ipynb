### one of the most common errorss in deep learning:shape errors

there are 2 rules performing matrix mutliplication needs to satisfy:
1. the **inner dimension** must match(A的行等于B的列)
* `(3,2) @ (3,2)` won't work
* `(2,3) @ (3,2)` will work
* `(3,2) @ (2,3)` will work

2. the resulting matrix has the shape of the **outer dimensions**（前矩阵第一行的列数）
* `(2,3) @ (3,2)` -> `(2,2)`
* `(3,2) @ (2,3)` -> `(3,3)`

* visit http://matrixmultiplication.xyz/ for more details


```python
torch.matmul(torch.rand(3,2),torch.rand(2,3))
```

## Reshaping,stacking,squeezing, and unsqueezing tensors

* reshaping: reshapes an input tensor to a defined shape
* View: Return a view of an input tensor of a certain shape but keep the same memory as the origional tensor
* stacking: combine multiple tensors on top of each other(vstack) or side by side(hstack)
* squeeze: remove all `1` dimensions from a tensor
* unsqueeze: add a `1`dimension to a target tensor
* Permute: return a view of the input with dimension permuted (swapped) in a certain way



```python
#creating
import torch 
x = torch.arange(1.,13.)
x,x.shape
```

```python
#add an extra dimension
x_reshaped = x.reshape(1,12)
x_reshaped
#reshape以后的tensor的参数量必须与reshape之前的tensor的参数量相同。
```

```python
#change the view
z = x.view(3,4)
z,z.shape
#z isn what x looks like they share the same memory
#so
```

```python
#changes Z changes X(because a view of tensor shares the same memory as the original input)
z[:,0] = 5
z,x
```

```python
#stack tensors on top of each other
x_stacked = torch.stack([x,x,x,x],dim=0)
x_stacked

```

```python
#squeeze_remove all single dimensions from tensor
#like:torch.size(2,1,2,1,2)->torch.size(2,2,2)
x_reshaped.squeeze().shape
#clear the 1
```

```python
x_squeezed = x_reshaped.squeeze()
#会发现，输出的结果少了一个括号。
x_squeezed
```

```python
print(f"Previous: {x_reshaped}")
print(f"  New   : {x_reshaped.squeeze()}")
```

```python
#unsqueeze: adds a single to a target tensor at a specific dim
print(f"Previous: {x_squeezed}")
print(f"  Now   : {x_squeezed.unsqueeze(dim=0)}")
#在dim=0上unsqueeze。也可以在dim=1上unsqueeze。
```

|名称	|ndim (阶)	|可用 dim|
|------|------|------|
|标量 (scalar)|	0	|无|
|向量 (vector)|	1|	0|
|矩阵 (matrix)	|2	|0, 1|
|3D张量	|3|	0, 1, 2|
|nD张量|	n|	0, 1, ..., n-1|

数学上，一个 `ndim=n` 的张量有 n 个轴 (axis) 或模 (mode)。每个轴编号从 0 到 n-1。`dim=k` 的意思是"沿着第 k 个轴进行缩减（reduction）"，操作后该轴被消除，张量的阶数减少 1。  

所以：   

* 一个 2D 矩阵 只有轴0（行方向）和轴1（列方向），没有轴2
* 操作 `sum(dim=2)` 相当于要求"在第3个轴的方向上求和"——但这个轴根本不存在

```python
#torch.permute: rearranges the dimensions of a target tensor in a specified order
x_original = torch.rand(size=(224,224,3))#height, width, color channels
#permute to rearrange the axis (or dim) order
x_permuted = x_original.permute(2,0,1)#shift axis original to now:0 to 1,1 to 2, 2 to 0

print(f"Previous:{x_original.shape}")
print(f"  Now   :{x_permuted.shape}")
```

```python
x_original[0,0,0] = 123211
x_original[0,0,0], x_permuted[0,0,0]
```

## Pytorch and numpy

numpy is a scientific Python numerical computing library   
because of this, Pytorch has functionality to intract with it
* data in numpy,want in pytorch tensor -> `torch.from_numpy(tensor)`
* Pytorch tensor -> Numpy ->`torch.Tensor.numpy()`

```python
#numpy array to tensor   
import torch
import numpy as np

array = np.arange(1.0,8.0)
tensor = torch.from_numpy(array)#converting from numpy ->pytorch, pytorch reflects numpy's default datatype of float64 unless specified otherwise 
array, tensor
```

```python
#change the valve of array, what will this do to tensor?
array = array + 1
array,tensor
```

```python
#tensor to numpy array
tensor = torch.ones(7)
numpy_tensor = tensor.numpy()
tensor, numpy_tensor
#default data type of pytorch is float 32
```

```python
#change the tensor, what happens to numpy_tensor?
tensor = tensor + 1
tensor, numpy_tensor
```

## Reproducbility (trying to take random out of random)

In short how a neural network learns:  
`start with random numbers -> tensor operations -> uodate random numbers to try and make them better representations of the data -> again ->again -> again`   

to reduce the randomness in nureon networks and pytorch comes the concept of a **random seed**   

Essentlly what the random seed does is "flavour" the randomness   


```python
torch.rand(3,3)
```

```python
import torch

#create two ranodm tensors
randomA = torch.rand(3,4)
randomB = torch.rand(3,4)
print(randomA)
print(randomB)
print(randomA == randomB)
```

```python
1 == 1

```

```python
#make some random but reproducible tensor
import torch

#set thr random seed
RANDOM_SEED = 123456
torch.manual_seed(RANDOM_SEED)
randomC = torch.rand(3,4)

torch.manual_seed(RANDOM_SEED)
randomD = torch.rand(3,4)

#every time reproduce a random seed
print(randomC == randomD)
```
